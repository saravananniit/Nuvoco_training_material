UTF-8 stands for:

Unicode Transformation Format - 8-bit

Breakdown
Unicode: A universal character encoding standard that assigns a unique code point to characters from virtually all languages and symbols.
Transformation Format: The method used to convert Unicode code points into bytes for storage and transmission.
8-bit: UTF-8 stores data using 8-bit (1-byte) units.

Key Features of UTF-8
Uses 1 to 4 bytes per character.
Compatible with ASCII (English characters use 1 byte).
Supports all Unicode characters, including:
    Tamil: வணக்கம்
    Hindi: नमस्ते
    Chinese: 你好
    Arabic: مرحبا
    Emoji: 😄
Examples
Character	Unicode	UTF-8   Bytes
A	        U+0041	        1 byte
₹	        U+20B9	        3 bytes
வ	       U+0BB5	       3 bytes
😄	        U+1F604	       4 bytes

```python
def write_to_file(filename: str) -> None:
    try:
        with open(filename, 'w', encoding='utf-8') as file:
            file.write("Writing to file!\n")
            file.write("Hello Good Morning!!\n")
        print("File written successfully.")

    except Exception as e:
        print(f"Error while writing: {e}")


write_to_file("task1.txt")
```

```python
def append_to_file(filename: str) -> None:
    try:
        with open(filename, 'a', encoding='utf-8') as file:
            file.write("Appending new content\n")
        print("Content appended successfully.")

    except Exception as e:
        print(f"Error while appending: {e}")


append_to_file("task.txt")
```

Structure - When it executes
except - Only if error occurs
else - Only if no error occurs
finally - Always executes
else + finally - Success logic + cleanup logic

Real‑world use (file, DB, API)

Rule Valid / Invalid:
else without except ❌ Invalid
finally without try ❌ Invalid
else after except   ✅ Valid
finally always last ✅ Required
try alone           ❌ Invalid

Python enforces strict ordering:
try → except → else → finally

```python
class FileProcessingError(Exception):  # creating user defined exception
    """Raise when there's read/write issue in a file."""
    pass


def read_numbers_from_file(file_path):
    """Raises custom errors if something fails."""
    try:
        with open(file_path, 'r') as f:
            lines = f.readlines()
            return [int(line.strip()) for line in lines]

    except FileNotFoundError:
        print("File not found. Please check the path.")

    except ValueError:
        raise FileProcessingError("Contains non-numeric data")

    except Exception as e:
        raise FileProcessingError(f"Unexpected error: {e}")

    finally:
        print("⚠ File read attempt completed.\n")
# Example usage
try:
    numbers = read_numbers_from_file("numbers.txt")
    print("Numbers read from file:", numbers)
    print(type(numbers))
    # convert this numbers to set
    numbers_set = set(numbers)
    print("Numbers as set:", numbers_set)
    # numbers to dictionary with value as square of number
    numbers_dict = {num: num**2 for num in numbers}
    print("Numbers as dictionary:", numbers_dict)
    # convert this numbers to list of strings
    numbers_str_list = [str(num) for num in numbers]
    print("Numbers as list of strings:", numbers_str_list)
    # convert this numbers to set comprehension with value as cube of number
    numbers_cube_set = {num**3 for num in numbers}
    print("Numbers as cube set:", numbers_cube_set)

except FileProcessingError as e:
    print(f"File processing error: {e}")

```

```python
import logging
from pathlib import Path

# Create logs directory
LOG_DIR = Path("logs")
LOG_DIR.mkdir(exist_ok=True)

# If the directory does not exist → ✅ create it
# If the directory already exists → ✅ do nothing   ✅ No exception is raised

# Log file path
LOG_FILE = LOG_DIR / "app.log"

# Create logger
logger = logging.getLogger("AppLogger")
logger.setLevel(logging.DEBUG)

# Create formatter
formatter = logging.Formatter(
    fmt="%(asctime)s | %(levelname)s | %(name)s | %(message)s",
    datefmt="%d-%m-%y %H:%M:%S"
)

# Console handler (INFO and above)
console_handler = logging.StreamHandler()
console_handler.setLevel(logging.INFO)
console_handler.setFormatter(formatter)

# File handler (DEBUG and above)
file_handler = logging.FileHandler(LOG_FILE)
file_handler.setLevel(logging.DEBUG)
file_handler.setFormatter(formatter)

# Avoid duplicate handlers
if not logger.handlers:
    logger.addHandler(console_handler)
    logger.addHandler(file_handler)

# Sample log messages
logger.debug("This is a DEBUG message")
logger.info("This is an INFO message")
logger.warning("This is a WARNING message")
logger.error("This is an ERROR message")
logger.critical("This is a CRITICAL message")
```
