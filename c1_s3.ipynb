```python
def calculate_area_of_circle():
    pi = 3.14159
    radius = 20
    area = pi * (radius ** 2)
    print(f"The circle's area: {area}")

calculate_area_of_circle()
```

```python
def calculate_area(length, width):
    area = length * width
    return area

result = calculate_area(5, 3)
print("The Rectangle's area is", result)
```

```python
# arguments are received like tuples

def greet(*args):
    for name in args:
        print(f"Hello, {name}!")

result=greet("Alice", 23, "Charlie","saravanan",90.8)
print(result)  # This will print None since greet() does not return anything
print(type(result))  # This will print <class 'NoneType'> since greet() does not return anything
```

```python
# creating flexible API
# inside the functions these arguments are stored as dict

def welcome(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

result=welcome(name="Alice", age=25, city="Delhi")
print(result)  # This will print None since welcome() does not return anything
print(type(result))  # This will print <class 'NoneType'> since welcome() does not return anything
```

```python
# Type hint: 

def add_numbers(a: int, b: int) -> int:
    return a + b

result = add_numbers(5, 10.5)
print(result)
print(type(result))

# Python is dynamically typed, type hints are not enforced at runtime, int + float results in a float, and runtime type enforcement requires tools like Pydantic.
```

```python
# concise and readable inline operations

# Regular Function
def square(x,y):
     return x * y

print(square(5,10))  

# functional programming style, using lambda functions for small operations
# Equivalent lambda function
square_lambda = lambda x,y: x * y

print(square_lambda(5,10))  
```

```python
# functional programming approach

# Using lambda with map()
numbers = [2, 4, 6, 8]

squares = tuple(map(lambda x: x * x, numbers))

print(squares)  
```

```python
# one line filtering conditions
# Using lambda with filter()
numbers = [5, 12, 7, 18, 3, 21]

result = list(filter(lambda x: x > 10, numbers))

print(result)  # Output: [12, 18, 21]
```

```python
# prime number check using lambda and filter
def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

prime_numbers = list(filter(is_prime, range(1, 21)))

print(prime_numbers)  # Output: [2, 3, 5, 7, 11, 13, 17, 19]

# use of lambda for prime number check
# prime_numbers_lambda = list(filter(lambda n: n > 1 and all(n % i != 0 for i in range(2, int(n**0.5) + 1)), range(1, 21)))   
# print(prime_numbers_lambda)  # Output: [2, 3, 5, 7, 11, 13, 17, 19]
```

```python
# local - function level
# global - program level

x = 10  # Global variable

def update_local():
    x = 5  # Local variable
    print("Inside update_local():", x)

def update_global():
    global x
    x = x + 5  # Modify global variable
    x=100
    print("Inside update_global():", x)

update_local()
print("After update_local(), x =", x)
update_global()
print("After update_global(), x =", x)
```

```python
# closure: powerful tool for creating reusable and stateful functions

# retain the state of variables from their enclosing scope even after the outer # function has finished execution

# doesnt rely on global variables

def counter():
    count = 0

    def increment():
        nonlocal count
        count += 1
        return count

    return increment

my_counter = counter()
print(my_counter())  # Output: 1
print(my_counter())  # Output: 2
print(my_counter())  # Output: 3
```
