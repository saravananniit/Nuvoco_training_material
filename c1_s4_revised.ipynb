Advantages of OOP:
Provides a clear structure to programs
Makes code easier to maintain, reuse, and debug
Helps keep your code DRY (Don't Repeat Yourself)

Tip: The DRY principle means you should avoid writing the same code more than once. Move repeated code into functions or classes and reuse it.

```python
class Dog:
    pass

print(Dog.__bases__)       # shows parent classes
# Output: (<class 'object'>,)  ← object is the parent!

print(Dog.__mro__)         # Method Resolution Order
# Output: [<class '__main__.Dog'>, <class 'object'>]

print(object.__bases__)
print(object.__mro__)

```

```python
# https://docs.python.org/3.12/library/random.html
import random


class Post:
    # Class variables
    total_posts = 0
    existing_ids = set()

    @classmethod
    def show_total_posts(cls):
        print(f"Total posts: {cls.total_posts}")

    # Static method to generate a unique ID
    @staticmethod
    def generate_post_id():
        while True:
            new_id = f"POST-{random.randint(1, 9999)}"
            if new_id not in Post.existing_ids:
                Post.existing_ids.add(new_id)
                return new_id
def main():
    posts = []

    # Create 5 posts
    for _ in range(5):
        post_id = Post.generate_post_id()
        posts.append(post_id)
        Post.total_posts += 1

    # Display generated post IDs
    print("Generated Post IDs:")
    for pid in posts:
        print(pid)

    # Show total posts
    Post.show_total_posts()


if __name__ == "__main__":
    main()
```

```python
class Post:
    def __init__(self, title):
        self._title = title

    @property
    def title(self):
        return self._title   # getter  method to get the attribute value

    @title.setter
    def title(self, value):
        self._title = value   # setter method to set the attribute value

    @title.deleter
    def title(self):
        # Optional: clear to a safe default before deleting
        # or directly delete the backing attribute.
        del self._title    # delete attribute


# --- minimal usage demo (optional) ---
post = Post("Hello World")
print(post.title)   # Hello World

post.title = "Updated Title"
print(post.title)   # Updated Title

del post.title
# Accessing post.title now will raise an AttributeError until it's set again
# print(post.title)   # Updated Title

post.title = "New One"  # set again if needed
```

```python
# https://pydantic.dev/docs/validation/latest/errors/validation_errors/#int_parsing
# pydantic-core is the package where the error handling system lives. 
# It bridges Rust validation logic with Python exception handling — 
# Rust validation failures produce ValError enums containing ValLineError instances, 
# which are then converted to Python ValidationError exceptions
# On GitHub (Original Rust Source)
# https://github.com/pydantic/pydantic-core/blob/main/src/errors/types.rs

from pydantic import BaseModel, ValidationError

class Model(BaseModel):
    x: int

try:
    result=Model(x="test")
    print(result.x)
except ValidationError as exc:
    print(exc.errors())
    print(exc.errors()[0]["msg"])



```

```python


# since validator is deperacted we have migrated to field_validator below
# -----------------------------------------------------------------------

from pydantic import BaseModel, field_validator, ValidationError
from datetime import date


class PostModel(BaseModel):
    title: str
    content: str
    date: date
    likes: int = 0
    comments: int = 0
    shares: int = 0

    # --- Field validators (Pydantic v2) ---

    @field_validator("title", mode="before")  
    @classmethod
    def title_must_not_be_empty(cls, v: str) -> str:
        # Strip whitespace and default to "Untitled" if empty
        return v.strip() or "Untitled"

    @field_validator("likes", "comments", "shares", mode="before")
    @classmethod
    def non_negative(cls, v: int) -> int:
        # Ensure non-negative counts
        return max(v, 0)


def main():
    # Case 1: Normal valid post
    post1 = PostModel(
        title="Morning Thoughts",
        content="Calm start to the day.",
        date=date(2025, 10, 12),
        likes=5,
        comments=2,
        shares=1,
    )

    # Case 2: Title is whitespace -> becomes "Untitled"
    post2 = PostModel(
        title="   ",
        content="Will be titled automatically.",
        date=date(2025, 10, 13),
    )

    # Case 3: Negative counts -> clamped to 0 by validator
    post3 = PostModel(
        title="Launch",
        content="Product demo",
        date=date(2025, 10, 13),
        likes=-10,
        comments=-2,
        shares=-5,
    )

    print("=== post1 ===")
    print(post1)
    print(post1.model_dump())

    print("\n=== post2 (title auto-fixed) ===")
    print(post2)
    print("Title ->", post2.title)

    print("\n=== post3 (counts clamped) ===")
    print(post3)
    print("likes, comments, shares ->", post3.likes, post3.comments, post3.shares)

    # Bonus: JSON serialization
    print("\n=== JSON ===")
    print(post1.model_dump_json())

    # Bonus: validation error example (missing required field)
    try:
        _ = PostModel(title="Oops", content="Missing date")
    except ValidationError as e:
        print("\n=== ValidationError example ===")
        print(e)


if __name__ == "__main__":
    main()

# | Mode     | When Runs         | Pydantic Validation Happens?   | Common Use              |
# | -------- | ----------------- | ----------------------------   | ----------------------- |
# | `before` | Before validation | ✅ Yes                        | Transform raw input     |
# | `after`  | After validation  | ✅ Already done               | Business-rule checks    |
# | `plain`  | Before validation | ❌ No                         | Fully custom validation |
# | `wrap`   | Around validation | ✅ Optional via handler       | Advanced control        |
```

# Pydantic validation flow (simplified)

1. Input data is received
2. mode="before" validators run
3. Pydantic converts types (e.g., str, int)
4. Normal validation happens
5. mode="after" validators run (default)
>The advantage of mode="before" is that it lets you clean, normalize, or fix bad input before Pydantic validates or converts it.

```python
from pydantic import BaseModel
from datetime import date


class PostModel(BaseModel):
    title: list[str]
    content: str
    date: date
    likes: int = 0


def main():
    post = PostModel(
        title=["Morning Thoughts"],
        content="Calm start to the day.",
        date=date(2025, 10, 12),
    )

    print(post)  # default __repr__

    print("\nAs Dictionary:")
    print(post.model_dump())

    print("\nAs JSON String:")
    print(post.model_dump_json(indent=4))


if __name__ == "__main__":
    main()
```

```python
# Polymorphism using duck typing (very Pythonic)
class Car:
    def move(self):
        print("Car is driving")

class Bike:
    def move(self):
        print("Bike is riding")

def travel(vehicle):
    vehicle.move()

travel(Car())
travel(Bike())
```

```python
# Polymorphism using method overriding 
class Animal:
    def make_sound(self):
        print("Animal makes a sound")

class Dog(Animal):
    def make_sound(self):
        print("Dog barks")

class Cat(Animal):
    def make_sound(self):
        print("Cat meows")

# Polymorphism in action
animals = [Dog(), Cat(), Animal()]

for animal in animals:
   animal.make_sound()
```

```python
from dataclasses import dataclass

# Version 1: Using @dataclass
@dataclass
class PostDataclass:
    title: str
    content: str

# Version 2: Manual implementation
class PostManual:
    def __init__(self, title, content):
        self.title = title
        self.content = content
    def __eq__(self, other):
        return self.title == other.title and self.content == other.content
    def __repr__(self):
        return f"Post(title={self.title!r}, content={self.content!r})"

# --- Create instances ---
p1 = PostDataclass("Hello", "World")
p2 = PostManual("Hello", "World")

# ✅ Prove __init__ works the same
print(p1.title, p1.content)   # Hello World
print(p2.title, p2.content)   # Hello World

# ✅ Prove __repr__ works the same
print(repr(p1))  # Post(title='Hello', content='World')
print(repr(p2))  # Post(title='Hello', content='World')

# ✅ Prove __eq__ works the same
a1 = PostDataclass("Hello", "World")
a2 = PostDataclass("Hello", "World")
b1 = PostManual("Hello", "World")
b2 = PostManual("Hello", "World")

print(a1 == a2)  # True  → compares field values, not memory address
print(b1 == b2)  # True  → same result

```
