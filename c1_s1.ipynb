```python
# Input: total purchase amount
amount = float(input("Enter your total amount: "))

# Decision-making with multiple conditions
if amount >= 5000:
    discount = 0.20 * amount
    print(f"You get a 20% discount of ₹{discount:.2f}.")
elif 2000 <= amount < 5000:
    discount = 0.10 * amount
    print(f"You get a 10% discount of ₹{discount:.2f}.")
else:
    loyaltyPoints = 100
    print(f"No discount, but you earn {loyaltyPoints} loyalty points.")

# Final message
print("Thank you for shopping with us!")
```

```python
# Secret code
secret_code = "treasure123"

# Ask user for the first guess
user_guess = input("Enter your guess: ")

# Keep asking until the guess is correct
while user_guess != secret_code:
    print("Wrong guess! Try again.")
    user_guess = input("Enter your guess: ")

# Correct guess
print("Congratulations! You've unlocked the treasure vault!")
```

```python
# Target number
target_num = 3

# Infinite loop
while True:
    guess_num = int(input("Guess a number: "))

    if guess_num == target_num:
        print("Congratulations!")
        break   # Exit the loop immediately

# Runs after the loop ends
print("Game Over!")
```

```python
# write a function to find factorial

def factorial(n):
    if n == 0:
        return 1
    else:
        return n * factorial(n - 1)
# Ask the user to enter a number
number = int(input("Enter a number to find its factorial: "))   
# Call the factorial function and print the result
result = factorial(number)
print(f"The factorial of {number} is {result}.")

```
