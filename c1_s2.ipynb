```python
# Attempting to modify a tuple (will raise TypeError)
car1 = ("Maruti Suzuki", "Swift", 3000)
# car1[2] = 3400   # ❌ This will raise: TypeError: 'tuple' object does not support item assignment

# Correct way: create a new tuple with updated value
car1 = ("Maruti Suzuki", "Swift", 3000)
car1 = (car1[0], car1[1], 3400)   # ✅ New tuple created

print(car1)
```

```python
# Packing: combining multiple values into a single tuple
car = ("Maruti Suzuki", "Swift", 3000)

# Unpacking: splitting tuple into individual variables
brand, model, price = car
print(brand, model, price)

# Packing multiple return values from a function
def get_car_details():
    car1 = ("Honda", "City", 3800)
    car2 = ("Hyundai", "Creta", 4000)
    return car1, car2

# Unpacking multiple returned tuples
car1, car2 = get_car_details()
print(car1)
print(car2)
```

```python
# Create set with default values
reg_numbers = {"MH12AB12", "MH12CD56", "MH12AB12", "MH12EF90"}

# print(reg_numbers)  # Output: {'MH12AB12', 'MH12CD56', 'MH12EF90'}
# Create an empty set
reg_numbers = set()

# Create set from a list using set() function
reg_numbers = set(["MH12AB12", "MH12CD56", "MH12AB12", "MH12EF90"])

print(reg_numbers)
```

```python
reg_numbers = {"MH12AB12", "MH12CD56", "MH12AB34", "MH12EF90"}

print(type(reg_numbers))  # Output: <class 'set'>
print(reg_numbers)  # Output: {'MH12AB12', 'MH12CD56', 'MH12AB34', 'MH12EF90'}
# Add new registration numbers
reg_numbers.add("MH12EF90")
print(reg_numbers)  # Output: {'MH12AB12', 'MH12CD56', 'MH12AB34', 'MH12EF90'}
# Remove a registration number
reg_numbers.remove("MH12AB34")
print(reg_numbers)  # Output: {'MH12AB12', 'MH12CD56', 'MH12EF90'}
# Display all unique registration numbers
for reg_number in reg_numbers:
    print(f"- {reg_number}")
```

```python
# update(): add all elements of another set
set1 = {1, 2, 3}
set2 = {3, 4, 5}
set1.update(set2)
print(set1)   # {1, 2, 3, 4, 5}

# pop(): removes and returns a random element
my_set = {"python", "java", "c++"}
print(my_set.pop())   # removes a first value


# discard(): safely remove an element without error
my_set = {1, 2, 3}
my_set.discard(5)   # No error even if value doesn't exist

# union(): combines elements from both sets
set1 = {1, 2, 3}
set2 = {3, 4, 5}
print(set1.union(set2))   # {1, 2, 3, 4, 5}

# intersection(): returns only common elements
set1 = {1, 2, 3}
set2 = {3, 4, 5}
print(set1.intersection(set2))   # {3}

```

```python
# Dictionary comprehension: mapping numbers to their squares
squares_dict = {x: x ** 2 for x in range(1, 6)}
print(squares_dict)   # {1: 1, 2: 4, 3: 9, 4: 16, 5: 25}

# List comprehension: list of squares
squares_list = [x ** 2 for x in range(1, 6)]
print(squares_list)   # [1, 4, 9, 16, 25]

# Set comprehension: unique squares (no duplicates)
squares_set = {x ** 2 for x in range(1, 6)}
print(squares_set)    # {1, 4, 9, 16, 25}

# tuple comprehension: generator expression (not a tuple)
squares_gen = (x ** 2 for x in range(1, 6))
print(squares_gen)    # <generator object <genexpr> at 0x...>
for square in squares_gen:
    print(square)  # Prints each square one by one
```

```python
# Using keys()
my_dict = {'name': 'John', 'age': 30}
print(my_dict.keys())   # dict_keys(['name', 'age'])

# Using values()
my_dict = {'name': 'John', 'age': 30}
print(my_dict.values())   # dict_values(['John', 30])

# Using popitem() — removes last inserted key-value pair
fruits = {"apple": 10, "banana": 20, "cherry": 15}
last_item = fruits.popitem()
print(last_item)   # ('cherry', 15)
print(fruits)      # {'apple': 10, 'banana': 20}

# Using clear() — removes all items
inventory = {"item1": 5, "item2": 8}
inventory.clear()
print(inventory)   # {}
```
